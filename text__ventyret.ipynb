```python
import time

print("Hej, vad kul att du har hittat hit!")
print("Det här programmet hjälper dig att hitta din fotbollsposition som passar just dig.")
print("Genom att svara på ett flertal frågor, kan jag hitta din inre fotbollsposition på nolltid.")
starta = input("Är du redo för frågorna? JA/NEJ?")

predicted_position = "Ingen position hittades."

if starta.lower() == "ja":
  print("Då sätter vi igång")
  fråga_1 = input("Känner du dig mer offensiv eller defensiv på fotbollsplanen?")

  if fråga_1.lower() == "offensiv":
    fråga_2 = input("\n{1}Jag är en målgörare som älskar att skjuta \n{2}Jag gillar att spela upp bollen och driva in i tomma ytor \n{3}Jag är en typisk passare som har vision för att hitta en bra pass \n{4}Jag gillar att göra inlägg och dribbla längst sidan\n Välj en av följande:")

    fråga_4 = input("Hur ser din kondition ut, bra/okej/dålig?")
    if fråga_4.lower() == "bra":
      print("Du har bra kondition")
    else:
      print("Kondition verkar inte vara din starkaste sida")

    if fråga_2 == "1":
      fråga_5 = input("Är du en fysisk spelare? Ja/Nej")
      if fråga_5.lower() == "ja":
        frååga_6 = int(input("Vad är din nuvarande vikt?"))
        if frååga_6 > 75:
          print("Låter bra, detta är en fördel på fotbollsplanen!")
        else:
          print("Det är helt lugnt")
        fråga_7 = input("Hur känner du kring hemåtjobbet? Bra/Dåligt?")
        if fråga_7.lower() == "bra":
          predicted_position = "offensiv mittfältare/falsk 9a"
          spelförståelse_passningsspel = input("Hur viktig är din spelförståelse och passningsförmåga för din roll? Viktigt/Mindre viktigt?")
          if spelförståelse_passningsspel.lower() == "viktigt": print("Mycket viktigt för en offensiv mittfältare!")
          rörelse_utan_boll = input("Är du bra på att hitta ytor och röra dig utan boll för att skapa chanser? JA/NEJ?")
          if rörelse_utan_boll.lower() == "ja": print("En nyckel för att bryta mönster!")
          initierar_anfall = input("Tycker du om att vara den som initierar anfall och är inblandad i de flesta målchanserna? JA/NEJ?")
          if initierar_anfall.lower() == "ja": print("Toppen, det är en viktig egenskap!")
          spela_rygg_mot_mål = input("Är du bekväm med att spela med ryggen mot mål och distribuera bollen till medspelare? JA/NEJ?")
          if spela_rygg_mot_mål.lower() == "ja": print("Utmärkt!")
        elif fråga_7.lower() == "dåligt":
          predicted_position = "central mittfältare"
          press_speluppbyggnad = input("Gillar du att vara delaktig i presspelet och den tidiga speluppbyggnaden? JA/NEJ?")
          if press_speluppbyggnad.lower() == "ja": print("Viktigt för att vinna boll högt!")
          taktisk_disciplin = input("Har du god taktisk disciplin och kan hålla din position väl? JA/NEJ?")
          if taktisk_disciplin.lower() == "ja": print("En grund för ett stabilt mittfält!")
          god_spelförståelse = input("Har du en god spelförståelse och förmåga att läsa spelet? JA/NEJ?")
          if god_spelförståelse.lower() == "ja": print("Det är viktigt för en central mittfältare!")
          länk_försvar_anfall = input("Gillar du att vara länken mellan försvar och anfall, och styra spelet centralt? JA/NEJ?")
          if länk_försvar_anfall.lower() == "ja": print("Låter perfekt!")
      elif fråga_5.lower() == "nej":
        print("Det är okej")
        predicted_position = "en snabb, teknisk forward"
        avslut_under_press = input("Är du effektiv i dina avslut även under hård press? JA/NEJ?")
        if avslut_under_press.lower() == "ja": print("Målskyttens signum!")
        djupledslöpningar = input("Gillar du att utmana försvarare med djupledslöpningar? JA/NEJ?")
        if djupledslöpningar.lower() == "ja": print("Skapar oreda i försvaret!")
        snabbhet_teknik = input("Litar du mer på din snabbhet och teknik än din fysik för att ta dig förbi försvarare? JA/NEJ?")
        if snabbhet_teknik.lower() == "ja": print("Din snabbhet är din styrka!")
        kontringar_öppna_ytor = input("Föredrar du att spela på kontringar och utnyttja öppna ytor i försvaret? JA/NEJ?")
        if kontringar_öppna_ytor.lower() == "ja": print("Det är en bra egenskap för en forward!")

    elif fråga_2 == "2":
      predicted_position = "yttermittfältare/nummer 10"
      avgörande_passningar = input("Är du bra på att leverera avgörande passningar i sista tredjedelen? JA/NEJ?")
      if avgörande_passningar.lower() == "ja": print("En riktig assistkung!")
      fria_roller = input("Föredrar du en fri roll där du kan röra dig över stora ytor? JA/NEJ?")
      if fri_roller.lower() == "ja": print("Kreativitet utan gränser!")
      bollkontroll_dribbla = input("Har du en utmärkt bollkontroll och förmåga att dribbla i trånga situationer? JA/NEJ?")
      if bollkontroll_dribbla.lower() == "ja": print("Teknik är nyckeln!")
      kreativ_spelare = input("Ser du dig själv som en kreativ spelare som kan låsa upp försvar med en oväntad passning? JA/NEJ?")
      if kreativ_spelare.lower() == "ja": print("En riktig playmaker!")
    elif fråga_2 == "3":
      predicted_position = "central mittfältare (spelfördelare)"
      tempo_kontroll = input("Har du förmågan att styra tempot i matchen med ditt passningsspel? JA/NEJ?")
      if tempo_kontroll.lower() == "ja": print("En viktig egenskap för en spelfördelare!")
      riskfyllda_passningar = input("Är du bekväm med att slå riskfyllda men genombrytande passningar? JA/NEJ?")
      if riskfyllda_passningar.lower() == "ja": print("Modig och effektiv!")
      lugn_med_boll = input("Är du lugn och trygg med bollen, även under press? JA/NEJ?")
      if lugn_med_boll.lower() == "ja": print("Kall som en filbunke!")
      långa_precisa_passningar = input("Har du en förmåga att slå långa, precisa passningar som bryter linjer? JA/NEJ?")
      if långa_precisa_passningar.lower() == "ja": print("En maestro med bollen!")
    elif fråga_2 == "4":
      predicted_position = "ytterforward/wing-back"
      utmana_en_mot_en = input("Gillar du att utmana din motståndare en-mot-en på kanten? JA/NEJ?")
      if utmana_en_mot_en.lower() == "ja": print("En dribbler som skapar ytor!")
      inlägg_precision = input("Har du precision i dina inlägg? JA/NEJ?")
      if inlägg_precision.lower() == "ja": print("Viktigt för att hitta anfallarna!")
      styrka_kanten_inlägg = input("Är din primära styrka att ta dig förbi din försvarare på kanten och slå inlägg? JA/NEJ?")
      if styrka_kanten_inlägg.lower() == "ja": print("Ett vasst vapen på kanten!")
      offensiv_defensiv_kapacitet = input("Har du både offensiva instinkter och arbetskapacitet för att hjälpa till i försvar? JA/NEJ?")
      if offensiv_defensiv_kapacitet.lower() == "ja": print("En komplett spelare!")

  elif fråga_1.lower() == "defensiv":
    fråga_3 = input("\n{1}Jag är en försvarsspelare som gillar att vinna tillbaks boll\n{2}Jag springer mycket och går all in på närkamper och dueller\n{3}Jag gillar att hålla mig i försvarslinjen hemma och sällan driva uppåt\n{4}Jag tycker om att vara både upp och nere på plan och stödja laget så mycket som möjligt \n Välj en av följande:")

    fråga_4 = input("Hur ser din kondition ut, bra/okej/dålig?")
    if fråga_4.lower() == "bra":
      print("Du har bra kondition")
    else:
      print("Kondition verkar inte vara din starkaste sida")

    if fråga_3 == "1":
      predicted_position = "defensiv mittfältare (bollvinnare)"
      bryta_anfall = input("Är din främsta uppgift att bryta motståndarnas anfall? JA/NEJ?")
      if bryta_anfall.lower() == "ja": print("En utmärkt defensiv förmåga!")
      enkel_bolldistribution = input("Föredrar du enkel och säker bolldistribution efter vunnen boll? JA/NEJ?")
      if enkel_bolldistribution.lower() == "ja": print("Effektivt och klokt spel!")
      kompromisslös_tacklingar = input("Är du en kompromisslös spelare som inte drar sig för tuffa tacklingar? JA/NEJ?")
      if kompromisslös_tacklingar.lower() == "ja": print("En riktig krigare på mittfältet!")
      positionera_bryta_passningar = input("Har du en stark förmåga att positionera dig för att bryta passningsvägar och vinna bollen? JA/NEJ?")
      if positionera_bryta_passningar.lower() == "ja": print("En smart bollvinnare!")
    elif fråga_3 == "2":
      predicted_position = "box-till-box mittfältare"
      understödja_anfall_försvar = input("Är du bra på att understödja både anfall och försvar effektivt? JA/NEJ?")
      if understödja_anfall_försvar.lower() == "ja": print("En oumbärlig spelare!")
      skott_utifrån = input("Har du ett bra skott utifrån straffområdet? JA/NEJ?")
      if skott_utifrån.lower() == "ja": print("Ett farligt vapen!")
      uthållighet_stora_ytor = input("Har du en exceptionell uthållighet och förmåga att täcka stora ytor på planen? JA/NEJ?")
      if uthållighet_stora_ytor.lower() == "ja": print("En motor på mittfältet!")
      inblandad_uppbyggnad_stöd = input("Gillar du att vara inblandad i både uppbyggnadsspelet och att stötta anfallet? JA/NEJ?")
      if inblandad_uppbyggnad_stöd.lower() == "ja": print("En komplett mittfältare!")
    elif fråga_3 == "3":
      predicted_position = "mittback/ytterback"
      placering_försvar = input("Har du god placeringsförmåga i försvaret? JA/NEJ?")
      if placering_försvar.lower() == "ja": print("En stabil grundbult!")
      uppspelsfas = input("Är du bekväm med att initiera uppspelsfasen från försvar? JA/NEJ?")
      if uppspelsfas.lower() == "ja": print("Viktigt för att bygga spelet bakifrån!")
      stark_i_luften_rensa_boll = input("Är du stark i luften och har en god förmåga att rensa bollen under press? JA/NEJ?")
      if stark_i_luften_rensa_boll.lower() == "ja": print("En klippa i försvaret!")
      ledare_försvar_kommunicera = input("Ser du dig själv som en ledare i försvaret som kommunicerar och organiserar backlinjen? JA/NEJ?")
      if ledare_försvar_kommunicera.lower() == "ja": print("En riktig general!")
    elif fråga_3 == "4":
      predicted_position = "wing-back/central mittfältare"
      mångsidighet_position = input("Kan du enkelt byta position och anpassa dig till nya roller under en match? JA/NEJ?")
      if mångsidighet_position.lower() == "ja": print("En flexibel tillgång!")
      balansspel = input("Är du bra på att hålla balansen i laget och fylla ytor? JA/NEJ?")
      if balansspel.lower() == "ja": print("Avgörande för lagets struktur!")
      kombination_kondition_snabbhet_teknik = input("Har du en bra kombination av kondition, snabbhet och teknisk skicklighet för att bidra över hela planen? JA/NEJ?")
      if kombination_kondition_snabbhet_teknik.lower() == "ja": print("En mångsidig spelare!")
      all_round_anpassa = input("Är du en 'all-round' spelare som kan anpassa sig till olika roller beroende på matchbilden? JA/NEJ?")
      if all_round_anpassa.lower() == "ja": print("Mycket användbar!")

  time.sleep(1)
  print("Din mest passande fotbollsposition är..")
  time.sleep(2)
  print(" trumvirvel")
  time.sleep(2)
  print(" duh duh")
  time.sleep(2)
  print(predicted_position)

  final_satisfaction = input("Är du nöjd med din tilldelade fotbollsposition? JA/NEJ?")
  if final_satisfaction.lower() == "ja":
      print("Underbart att höra!")
  else:
      print("Okej, vi hoppas du hittar den perfekta positionen ändå!")

  review = input("Vänligen betygsätt programmet (1-5 stjärnor): ")
  review_int = int(review)
  if 1 <= review_int <= 5:
      print(f"Tack för ditt betyg på {review_int} stjärnor!")
  else:
      print("Betyget måste vara mellan 1 och 5. Tack ändå!")

else:
  print("Ok, då avbryter vi programmet.")
```
