```python
import random

vunna_rundor = 0
förlorade_rundor = 0

while vunna_rundor < 2 and förlorade_rundor < 2:
    antal_rundor = vunna_rundor + förlorade_rundor

    if antal_rundor == 0:
        print("Första rundan här, nu drar vi igång med spelet!")
    else:
        print(f"Ny runda, jalla! Du har vunnit {vunna_rundor} rundor och datorn {förlorade_rundor}")

    random_number = random.randint(1, 10)
    vann_rundan = False

    for försök in range(1, 4):
        guess = int(input(f"Försök {försök}/3 - Gissa ett tal mellan 1 och 10: "))

        if guess < random_number:
            print("Du gissade för lågt.")
        elif guess > random_number:
            print("Du gissade för högt.")
        else:
            print("Rätt! Du vann denna runda!")
            vann_rundan = True
            vunna_rundor = vunna_rundor + 1
            break

    if vann_rundan == False:
        print(f"Tyvärr, försöken är slut. Det rätta talet var {random_number}.")
        förlorade_rundor = förlorade_rundor + 1

if vunna_rundor == 2:
    print("Grattis! Du vann 2 rundor och tog hem hela spelet!")
else:
    print("Datorn vann 2 rundor. Bättre lycka nästa gång!")
```
