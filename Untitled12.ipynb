```python
saldo = 1250

while True:
    print("\n=== Välkommen till systemet ===")
    print("1. Visa saldo")
    print("2. Sätt in pengar")
    print("3. Ta ut pengar")
    print("4. Avsluta")

    val = input("Välj ett alternativ (1-4): ")

    if val == "1":
        print(f"Ditt saldo är: {saldo} kr")
        break

    elif val == "2":
        sätt_in = int(input("Hur mycket vill du sätta in? "))
        saldo += sätt_in
        print(f"Din nya saldo är: {saldo} kr")
        break

    elif val == "3":
        ta_ut = int(input("Hur mycket vill du ta ut? "))
        saldo -= ta_ut
        print(f"Din nya saldo är: {saldo} kr")
        break

    elif val == "4":
        print("Hej då!")
        break

    else:
        print("Ogiltigt val. Välj ett alternativ mellan 1 och 4.")
```
