```python
import random

datorns_tal = random.randint(1, 1000)
print("Jag tänker på ett nummer mellan 1 och 1000.")

antal_försök = 10
korrekt_gissat = False


for försök_nr in range(1, antal_försök + 1):
  fråga = int(input(f"Försök {försök_nr}/{antal_försök}: Vilket tal tror du datorn tänker på? "))

  if fråga == datorns_tal:
    print(f"Grattis din gissning är helt korrekt! Du gissade rätt på {försök_nr} försök.")
    korrekt_gissat = True


  if fråga < datorns_tal:
    print(f"Kom igen nu kungen, Högre än {fråga} kan du!")

  elif fråga == datorns_tal:
    print(f"Grattis din gissning är helt korrekt! Du gissade rätt på {försök_nr} försök.")
    korrekt_gissat = True

  else:
    print(f"Ojoj vad stort du gick, lägre än {fråga} kan du nog!")

if not korrekt_gissat:
  print(f"Tyvärr, du tog slut på försök. Det rätta talet var {datorns_tal}.")
```
