```python
import random
import time

spelare_namn = input("Ange namn för spelare 1: ")
print(f"Hej {spelare_namn}!")

poang = 0

starta_spel = input("Är du redo att starta spelet? (Skriv vad som helst för att köra, eller 'nej' för att avsluta): ")
if starta_spel.lower() != "nej":
    print("Så kul, då startar vi genast!")

    time.sleep(1)
    print("Spelet går ut på att du först möter en slumpad motståndare.")

    time.sleep(2)
    print("Därefter får du slumpade krafter som skadar olika mycket.")

    time.sleep(3)
    print("Om du förlorar får du börja om spelet igen. Du får minus 1 poäng för varje förlust, och målet är att nå 3 poäng för att vinna!")

    time.sleep(4)
    print("OBS! Under rundorna kommer du även att få svara på en snabb mattefråga för att öka din skada ytterligare!")

    time.sleep(3)
    print("OBS! Du kan inte få minuspoäng, den lägsta möjliga poängen är 0.")

    time.sleep(2)

    krafter = [
        "Eldklot",
        "Trästock",
        "Raket",
        "Tromb",
        "Pilbåge",
        "Snöboll",
        "Blixtnedslag",
        "Djupfrysning",
        "Elchock",
        "Giftmoln",
        "Jordskalv"
    ]

    skade_tabell = {
        "Eldklot": 50,
        "Trästock": 30,
        "Raket": 90,
        "Tromb": 40,
        "Pilbåge": 25,
        "Snöboll": 20,
        "Blixtnedslag": 80,
        "Djupfrysning": 15,
        "Elchock": 10,
        "Giftmoln": 45,
        "Jordskalv": 60
    }

    spela_vidare = True
    while spela_vidare:

        motstandar_val = random.randint(1, 6)
        motstandare_namn = ""
        if motstandar_val == 1:
            motstandare_namn = "Master Maximilian"
        elif motstandar_val == 2:
            motstandare_namn = "Evil Solicus"
        elif motstandar_val == 3:
            motstandare_namn = "Lolipop Robin"
        elif motstandar_val == 4:
            motstandare_namn = "Clownen Jens"
        elif motstandar_val == 5:
            motstandare_namn = "Bävern Majki"
        elif motstandar_val == 6:
            motstandare_namn = "JonatHAN Genius"

        print(f"\n NY RUNDA!! ")
        print(f"Din motståndare denna runda är: {motstandare_namn}")
        print(f"Nuvarande poäng: {poang}/3")
        time.sleep(2)

        egen_kraft = random.choice(krafter)
        print(f"\nDu fick kraften: {egen_kraft} (Grundskada: {skade_tabell[egen_kraft]})")
        time.sleep(2)

        element = input("Välj element-fokus för din kraft (Skriv 'eld', 'is' eller 'blixt'): ").lower()
        element_bonus = 0
        if element == "eld":
            element_bonus = 5
            print("Eldfokus! Ger +5 i brännskada.")
        elif element == "is":
            element_bonus = 5
            print("Isfokus! Fryser motståndaren lite grann och ger +5 skada.")
        elif element == "blixt":
            element_bonus = 5
            print("Blixtfokus nudå! Chockar motståndaren och ger +5 skada.")
        else:
            print("Inget giltigt fokus valt, ingen extra elementbonus denna gång.")

        time.sleep(2)
        strategi_val = input("Vill du spela defensivt eller offensivt? (Skriv 'offensivt' eller 'defensivt'): ").lower()

        offensiv_bonus = 0
        defensiv_reducering = 0

        if strategi_val == "offensiv":
            offensiv_bonus = 15
            print("Du väljer att satsa allt framåt! Du gör +15 extra skada!")
        elif strategi_val == "defensiv":
            defensiv_reducering = 15
            print("Du skyddar dig! Du tar 15 mindre skada denna runda!")
        else:
            print("Du kunde inte bestämma dig, så du kör som vanligt!")

        time.sleep(2)

        tal_1 = random.randint(1, 10000)
        tal_2 = random.randint(1, 10000)
        ratt_svar = tal_1 + tal_2

        print(f"\n MATTEUTMANINGEN")
        print(f"Svara rätt på frågan för att göra ytterligare +10 skada!")
        svar = input(f"Vad är {tal_1} + {tal_2}? ")

        matte_bonus = 0

        if svar == str(ratt_svar):
            matte_bonus = 10
            print("Snyggt! Rätt svar! Du får +10 i Super-Boost!")
        else:
            print(f"Tyvärr var det fel svar (rätt svar var {ratt_svar}). Ingen extra boost denna runda.")

        time.sleep(2)
        print(f"\nDu kastar din {egen_kraft} på {motstandare_namn}!")
        time.sleep(2)

        motstandar_kraft = random.choice(krafter)
        print(f"{motstandare_namn} fick kraften: {motstandar_kraft} (Grundskada: {skade_tabell[motstandar_kraft]})")
        time.sleep(2)

        print(f"{motstandare_namn} kastar {motstandar_kraft} på dig!")
        time.sleep(2)

        print("\nNu tar ni skada")
        time.sleep(1)

        egen_skada = skade_tabell[motstandar_kraft] - defensiv_reducering
        if egen_skada < 0:
            egen_skada = 0

        motstandar_skada = skade_tabell[egen_kraft] + offensiv_bonus + element_bonus + matte_bonus

        print(f"{spelare_namn} tog {egen_skada} i skada!")
        print(f"{motstandare_namn} tog {motstandar_skada} i skada!")
        time.sleep(2)

        if motstandar_skada > egen_skada:
            print(f"\nSnyggt! Du gjorde mest skada och vann denna runda!")
            poang += 1
        elif egen_skada > motstandar_skada:
            print(f"\nAj då! {motstandare_namn} vann denna runda.")
            poang -= 1
            if poang < 0:
                poang = 0
        else:
            print("\nDet blev oavgjort, båda tog lika mycket skada!")

        print(f"Dina poäng efter denna runda: {poang}/3")
        time.sleep(2)

        if poang >= 3:
            print(f"\nGRATTIS {spelare_namn}! Du har nått 3 poäng och vunnit hela spelet!")
            break

        val_fortsatt = input("\nVill du fortsätta spela nästa runda? (Skriv 'nej' för att avsluta, eller tryck enter för att fortsätta): ")
        if val_fortsatt.lower() == "nej":
            spela_vidare = False
            print("\nTack för att du spelade!")

else:
    print("Okej, då avbryter vi programmet.")
```

Varför använde du just en while-loop istället för en for-loop?

Det som avgjorde att jag behövde använda en while-loop och inte en for-loop i detta program var för att med en for - loop (for i in range(x)måste vi på förhand bestämma exakt hur många gånger loopen ska köras, till exempel exakt 3.

Men i detta spel är målet att nå 3 poäng för att vinna, men vi kan även förlora och gå minus en poäng för varje förlorad runda.
