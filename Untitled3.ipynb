```python
tal_1 = float(input("Skriv in ditt första tal:"))
tal_2 = float(input("Skriv in ditt andra tal:"))

summa = tal_1 + tal_2
differens = tal_1 - tal_2
produkt = tal_1 * tal_2

print(f"Summan är: {summa}")
print(f"Differensen är: {differens}")
print(f"Produkten är: {produkt}")
```
