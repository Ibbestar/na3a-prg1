```python
alfabet = "abcdefghijklmnopqrstuvwxyzåäö"
meddelande = "hej"
nyckel = 25

krypterat_meddelande = ""
#UPPDRAG: Skriv en loop som skriver ut "ifk"
#Tips: Använd .index() och []
for bokstav in meddelande:
  plats = alfabet.index(bokstav)
  ny_plats = (plats + nyckel) % len(alfabet)
  krypterat_meddelande += alfabet[ny_plats]

print(krypterat_meddelande)
```
