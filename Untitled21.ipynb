```python
alfabet = "abcdefghijklmnopqrstuvwxyzåäö"
meddelande = "khm"


for nyckel in range(len(alfabet)):
    inkrypterat = ""
    for bokstav in meddelande:
        if bokstav in alfabet:
            ny_plats = (alfabet.index(bokstav) - nyckel) % len(alfabet)
            inkrypterat += alfabet[ny_plats]
        else:
            inkrypterat += bokstav
    print(f"Nyckel {nyckel}: {inkrypterat}")
```
